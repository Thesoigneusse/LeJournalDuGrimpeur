## Extraction de la cotation

```python
from typing import List

def extract_cotation(cot: List[str]) -> 'Cotation':
    from Classes.Cotation import Cotation
    test = list(set(cot) - set(['Cotation', 'globale', 'libre', 'obligatoire', 'artificielle', 'Engagement', 'Qualité', 'de', "l'équipement", 'Exposition', 'rocher']))
    for i in range(len(test)):
        print(f"{i}: {test[i]}")

    temp = {}

    for i in range(len(test)):
        value = test[i].upper()
        if value in Cotation.possible_artif:
            temp["artif"] = value
        elif value in Cotation.possible_engag:
            temp["engag"] = value
        elif value in Cotation.possible_risqu:
            temp["risqu"] = value
        elif value in Cotation.possible_equip:
            temp["equip"] = value
        elif value in Cotation.possible_expos:
            temp["expos"] = value
        elif value in Cotation.possible_globa:
            temp["globa"] = value
        elif value in Cotation.possible_libre:
            if "libre" in temp.keys():
                if Cotation.possible_libre.index(temp['libre']) < Cotation.possible_libre.index(value):
                    temp["oblig"] = temp['libre']
                    temp['libre'] = value
                else:
                    temp["oblig"] = value
            else:
                temp["libre"] = value
        

        
        
print(f"{temp}")
print(Cotation(libre=temp['libre'] if 'libre' in temp.keys() else None,
                globale=temp["globa"] if 'globa' in temp.keys() else None,
                obligatoire=temp["oblig"] if 'oblig' in temp.keys() else None,
                artificielle=temp["artif"] if 'artif' in temp.keys() else None,
                engagement=temp["engag"] if 'engag' in temp.keys() else None,
                risque=temp["risqu"] if 'risqu' in temp.keys() else None,
                equipement=temp["equip"] if 'equip' in temp.keys() else None,
                exposition=temp["expos"] if 'expo' in temp.keys() else None))
```
